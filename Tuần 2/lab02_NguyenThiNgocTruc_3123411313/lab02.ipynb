### Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
import torch

a = torch.rand(3, 4)    # phan phoi deu U(0, 1)
b = torch.randn(3, 4)   # phan phoi chuan N(0, 1)
c = torch.zeros(2, 3)   # toan so 0
d = torch.ones(2, 3)    # toan so 1
e = torch.eye(4)        # ma tran don vi 4x4

print("a = torch.rand(3,4):");  print(a)
print("\nb = torch.randn(3,4):"); print(b)
print("\nc = torch.zeros(2,3):"); print(c)
print("\nd = torch.ones(2,3):");  print(d)
print("\ne = torch.eye(4):");     print(e)
```

**Yêu cầu 1:** Nhận xét sự khác nhau giữa `torch.rand()` và `torch.randn()`.

```python
r1 = torch.rand(10000)
r2 = torch.randn(10000)
print("torch.rand : min=%.4f  max=%.4f  mean=%.4f  std=%.4f" % (r1.min(), r1.max(), r1.mean(), r1.std()))
print("torch.randn: min=%.4f  max=%.4f  mean=%.4f  std=%.4f" % (r2.min(), r2.max(), r2.mean(), r2.std()))
```

**Nhận xét:**

- `torch.randn()` sinh số theo **phân phối chuẩn (Gauss) N(0, 1)**: giá trị có thể âm hoặc dương,
  không bị chặn, trung bình xấp xỉ 0 và độ lệch chuẩn xấp xỉ 1.
- Trong học sâu, `randn` thường được dùng để khởi tạo trọng số vì phân phối đối xứng quanh 0.

Sử dụng `shape`, `size()`, `ndim` và `dtype` để kiểm tra các thuộc tính của Tensor.

```python
import torch

a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("\nShape              :", a.shape)
print("Size               :", a.size())
print("Number of dimensions:", a.ndim)
print("Data type          :", a.dtype)
print("Number of elements :", a.numel())
```

- Chiều 0 (dim=0) có 2 phần tử → 2 "khối" (block/batch).
- Chiều 1 (dim=1) có 3 phần tử → mỗi khối là ma trận 3 hàng.
- Chiều 2 (dim=2) có 4 phần tử → mỗi hàng có 4 cột.

nên có thể truy xuất như tuple: `a.shape[0] = 2`. Ngoài ra `a.shape` và `a.size()` cho kết quả giống hệt nhau,
chỉ khác `shape` là thuộc tính còn `size()` là phương thức.


### Bài 4. Slicing và Indexing
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
import torch

a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

**Yêu cầu 1:** Cho biết shape của `a[1]` và `a[1:, 2:4]`.

```python
print("shape cua a          :", a.shape)
print("shape cua a[1]       :", a[1].shape)
print("shape cua a[1:, 2:4] :", a[1:, 2:4].shape)
```

**Nhận xét:**
- `a[1]` là **indexing** trên chiều 0 → chiều này bị "nuốt" mất, kết quả có shape `[4, 5]` (giảm 1 chiều).
- `a[1:, 2:4]` là **slicing**: chiều 0 lấy từ chỉ số 1 đến hết → còn 2 phần tử; chiều 1 lấy chỉ số 2 và 3 →
  còn 2 phần tử; chiều 2 giữ nguyên 5 → shape `[2, 2, 5]` (vẫn 3 chiều).
- Điểm mấu chốt: **index bằng số nguyên làm giảm số chiều, còn slice `:` thì giữ nguyên số chiều.**

**Yêu cầu 2:** Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

```python
last2 = a[:, :, -2:]
print("a[:, :, -2:] shape:", last2.shape)
print(last2)
```

**Nhận xét:** `a[:, :, -2:]` giữ nguyên chiều 0 và chiều 1, chỉ lấy 2 phần tử cuối của chiều 2,
nên shape từ `[3, 4, 5]` trở thành `[3, 4, 2]`.


### Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
import torch

a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

**Yêu cầu 1:** Kiểm tra số phần tử trước và sau khi reshape.

```python
b = a.reshape(3, 2, 2)
print("Truoc reshape: shape =", a.shape, "| so phan tu =", a.numel())
print("Sau  reshape : shape =", b.shape, "| so phan tu =", b.numel())
print("So phan tu co bang nhau khong?", a.numel() == b.numel())
```

**Yêu cầu 2:** Thử reshape Tensor thành kích thước 2 × 6.

```python
c = a.reshape(2, 6)
print("a.reshape(2, 6):")
print(c)
print("shape:", c.shape, "| numel:", c.numel())

# Dung -1 de PyTorch tu suy ra so chieu con lai
print("\na.reshape(-1, 6).shape =", a.reshape(-1, 6).shape)
print("a.reshape(2, -1).shape  =", a.reshape(2, -1).shape)

# Thu reshape sai kich thuoc de thay loi
try:
    a.reshape(5, 3)
except RuntimeError as err:
    print("\nLoi khi reshape(5,3):", err)
```

**Nhận xét:**
- `ravel()` (hoặc `flatten()`) trải tensor thành vector 1 chiều 12 phần tử.
- Reshape chỉ thay đổi **cách diễn giải** bộ nhớ, **không thay đổi dữ liệu và số phần tử**:
  3×4 = 3×2×2 = 2×6 = 12 phần tử.
- Nếu tích các chiều mới khác 12 (ví dụ 5×3 = 15) thì PyTorch báo lỗi.
- Có thể dùng `-1` để PyTorch tự tính một chiều còn lại.


### Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử `@`.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:"); print(a)
print("\nb:"); print(b)
print("\na + b:"); print(a + b)
print("\na / b:"); print(a / b)
print("\na^2:");   print(a ** 2)
print("\nMatrix multiplication (a @ b.T):"); print(a @ b.T)
```

**Yêu cầu:** Phân biệt phép toán `a * b` với `a @ b.T`; cho biết shape của kết quả `a @ b.T`.

```python
mul_elem = a * b        # nhan tung phan tu (element-wise / Hadamard)
mul_mat  = a @ b.T      # nhan ma tran

print("a * b   (element-wise) shape:", mul_elem.shape)
print(mul_elem)
print("\na @ b.T (matrix mul)  shape:", mul_mat.shape)
print(mul_mat)

# Kiem chung: phan tu (0,0) cua a@b.T chinh la tich vo huong hang 0 cua a va hang 0 cua b
print("\nKiem chung (a@b.T)[0,0] =", (a[0] * b[0]).sum().item(), "vs", mul_mat[0, 0].item())
```

**Nhận xét:**
- `a * b` là phép nhân **từng phần tử** (Hadamard). Hai tensor phải cùng shape (hoặc broadcast được);
  kết quả có shape `[2, 3]` giống đầu vào.
- `a @ b.T` là phép **nhân ma trận**: `a` có shape (2, 3), `b.T` có shape (3, 2), điều kiện nhân là số cột của
  ma trận trái bằng số hàng của ma trận phải (3 = 3). Kết quả có shape **`[2, 2]`**.
- Phần tử `(i, j)` của `a @ b.T` là tích vô hướng giữa hàng `i` của `a` và hàng `j` của `b`.
- Không thể viết `a @ b` vì (2,3) @ (2,3) không thoả điều kiện nhân ma trận.


### Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch

a = torch.randn(3, 4)
print(a)
print("\nMean (dim=0):");              print(torch.mean(a, dim=0))
print("\nStandard deviation (dim=0):"); print(torch.std(a, dim=0))
print("\nCumulative sum (dim=0):");     print(torch.cumsum(a, dim=0))
```

**Yêu cầu:** Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

```python
print("Mean (dim=1):");               print(torch.mean(a, dim=1))
print("\nStandard deviation (dim=1):"); print(torch.std(a, dim=1))
print("\nCumulative sum (dim=1):");     print(torch.cumsum(a, dim=1))

print("\nShape ket qua:")
print("  mean(dim=0):", torch.mean(a, dim=0).shape, "  mean(dim=1):", torch.mean(a, dim=1).shape)
print("  cumsum(dim=0):", torch.cumsum(a, dim=0).shape, "cumsum(dim=1):", torch.cumsum(a, dim=1).shape)
```

**Nhận xét:**
- `dim` là chiều bị **thu gọn (reduce)**. Với tensor `a` shape (3, 4):
  - `dim=0` → gộp theo hàng, tính **theo từng cột**, kết quả là vector 4 phần tử (shape `[4]`).
  - `dim=1` → gộp theo cột, tính **theo từng hàng**, kết quả là vector 3 phần tử (shape `[3]`).
- Riêng `cumsum` không làm giảm số chiều mà chỉ cộng dồn dọc theo chiều được chỉ định, nên kết quả
  vẫn giữ shape `[3, 4]`: `dim=0` cộng dồn từ trên xuống, `dim=1` cộng dồn từ trái sang phải.
- Lưu ý: `torch.std` mặc định dùng ước lượng không chệch (chia cho n-1).


### Bài 8. Autograd cơ bản
Tạo biến có `requires_grad=True` và dùng `backward()` để tính đạo hàm tự động.

```python
import torch

x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

**Yêu cầu 1:** Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với `x.grad`.

```python
manual = 2 * 3.6          # dy/dx = 2x  ->  2 * 3.6 = 7.2
print("Dao ham thu cong dy/dx = 2x = 2 * 3.6 =", manual)
print("Autograd  x.grad       =", x.grad.item())
print("Bang nhau?", abs(manual - x.grad.item()) < 1e-6)
```

**Yêu cầu 2:** Đổi hàm thành y = 3x² + 2x + 1 và kiểm tra gradient.

```python
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()

manual2 = 6 * 3.6 + 2     # dy/dx = 6x + 2
print("y  =", y2.item())
print("Dao ham thu cong dy/dx = 6x + 2 =", manual2)
print("Autograd  x2.grad              =", x2.grad.item())
print("Bang nhau?", abs(manual2 - x2.grad.item()) < 1e-5)
```

**Nhận xét:**
- Với y = x², đạo hàm là 2x; tại x = 3.6 cho 7.2 — trùng khớp với `x.grad`.
- Với y = 3x² + 2x + 1, đạo hàm là 6x + 2; tại x = 3.6 cho 23.6 — cũng trùng khớp.
- PyTorch xây dựng **đồ thị tính toán động** khi thực hiện các phép toán trên tensor có
  `requires_grad=True`. Khi gọi `backward()`, nó lan truyền ngược (chain rule) và cộng dồn kết quả vào `.grad`.
- Lưu ý: `.grad` được **cộng dồn**, nên trong vòng lặp huấn luyện phải gọi `zero_grad()` trước mỗi bước.


### Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])          # y = x^2 + 2x + 3
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])   # ma tran dac trung [x^2, x, 1]

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()              # xoa gradient cu
    y_pred = x @ w                     # du doan
    mse = torch.mean(torch.square(y - y_pred))   # ham mat mat
    mse.backward()                     # lan truyen nguoc, tinh gradient
    optimizer.step()                   # cap nhat tham so
    if (epoch + 1) % 200 == 0:
        print("Epoch %4d | MSE = %.6f" % (epoch + 1, mse.item()))

print("\nLearned coefficients:")
print(w)
```

**Yêu cầu 1:** So sánh ba hệ số học được với [1, 2, 3].

```python
learned = w.detach().numpy().ravel()
true    = np.array([1.0, 2.0, 3.0])
names   = ["a (he so x^2)", "b (he so x)", "c (hang so)"]

print("%-16s %12s %12s %12s" % ("He so", "That", "Hoc duoc", "Sai so"))
for n, t, l in zip(names, true, learned):
    print("%-16s %12.4f %12.4f %12.6f" % (n, t, l, abs(t - l)))

print("\nMSE cuoi cung:", mse.item())
```

**Nhận xét:** Ba hệ số học được xấp xỉ [1, 2, 3] với sai số rất nhỏ. Sai số còn lại là do số bước lặp hữu hạn
(1000 epoch) và learning rate; nếu tăng số epoch hoặc điều chỉnh `lr` thì hệ số sẽ càng chính xác.
Vì dữ liệu được sinh ra không có nhiễu và mô hình tuyến tính theo tham số, nghiệm tối ưu chính là [1, 2, 3].

**Yêu cầu 2:** Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.

**Trả lời:**

| Hàm | Vai trò |
|---|---|
| `optimizer.zero_grad()` | Xoá (đặt về 0) gradient của các tham số từ bước lặp trước. PyTorch **cộng dồn** gradient vào `.grad`, nếu không xoá thì gradient các vòng lặp sẽ bị cộng chồng lên nhau và làm hỏng quá trình huấn luyện. |
| `mse.backward()` | Lan truyền ngược trên đồ thị tính toán, áp dụng quy tắc chuỗi để tính ∂MSE/∂w và lưu vào `w.grad`. |
| `optimizer.step()` | Dùng gradient vừa tính để **cập nhật tham số** theo thuật toán tối ưu (ở đây là NAdam - Adam kết hợp Nesterov momentum). |

Thứ tự bắt buộc là: `zero_grad()` → tính loss → `backward()` → `step()`.


### Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình:
A + B = 9, C − D = 1, A + C = 8, B − D = 2.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for epoch in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
    if (epoch + 1) % 500 == 0:
        print("Epoch %4d | sqerr = %.8f" % (epoch + 1, sqerr.item()))

print("\nA =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

**Yêu cầu 1:** Kiểm tra nghiệm thu được với bốn phương trình.

```python
a_, b_, c_, d_ = A.item(), B.item(), C.item(), D.item()
print("A = %.6f, B = %.6f, C = %.6f, D = %.6f\n" % (a_, b_, c_, d_))

checks = [("A + B = 9", a_ + b_, 9),
          ("C - D = 1", c_ - d_, 1),
          ("A + C = 8", a_ + c_, 8),
          ("B - D = 2", b_ - d_, 2)]

print("%-12s %12s %10s %12s" % ("Phuong trinh", "Gia tri tinh", "Ve phai", "Sai so"))
for name, val, rhs in checks:
    print("%-12s %12.6f %10d %12.6f" % (name, val, rhs, abs(val - rhs)))
```

**Yêu cầu 2:** Nhận xét vì sao bài toán có thể có nhiều nghiệm.

**Trả lời:** Hệ có 4 phương trình và 4 ẩn, nhưng các phương trình **không độc lập tuyến tính**. Cụ thể:

(A + B) − (C − D) = 9 − 1 = 8, mà (A + C) − (B − D) = 8 − 2 = 6...

Xét kỹ hơn: từ pt1: B = 9 − A; từ pt3: C = 8 − A; thay vào pt2: D = C − 1 = 7 − A;
kiểm tra pt4: B − D = (9 − A) − (7 − A) = 2, đúng bằng vế phải — pt4 **tự động thoả mãn** với mọi A.

Vậy phương trình thứ 4 là **hệ quả của ba phương trình đầu** (hạng của ma trận hệ số chỉ bằng 3 < 4),
nên hệ có **vô số nghiệm** phụ thuộc một tham số tự do A:
> A = t, B = 9 − t, C = 8 − t, D = 7 − t (với t bất kỳ).

Do đó nghiệm mà autograd tìm được phụ thuộc vào **giá trị khởi tạo ngẫu nhiên**: mỗi lần chạy sẽ cho một
bộ (A, B, C, D) khác nhau, nhưng tất cả đều thoả mãn cả bốn phương trình (sqerr ≈ 0) và đều thoả
A + B = 9, B = 9 − A, C = 8 − A, D = 7 − A.

```python
# Kiem chung: nghiem thu duoc co dung dang (t, 9-t, 8-t, 7-t) khong?
t = A.item()
print("t        =", round(t, 6))
print("9 - t    =", round(9 - t, 6), " | B =", round(B.item(), 6))
print("8 - t    =", round(8 - t, 6), " | C =", round(C.item(), 6))
print("7 - t    =", round(7 - t, 6), " | D =", round(D.item(), 6))

# Chay lai voi khoi tao khac de thay nghiem khac
A2 = torch.tensor(50.0, requires_grad=True)
B2 = torch.tensor(-3.0, requires_grad=True)
C2 = torch.tensor(0.0,  requires_grad=True)
D2 = torch.tensor(10.0, requires_grad=True)
opt2 = torch.optim.NAdam([A2, B2, C2, D2], lr=0.05)
for _ in range(5000):
    e = (A2+B2-9)**2 + (C2-D2-1)**2 + (A2+C2-8)**2 + (B2-D2-2)**2
    opt2.zero_grad(); e.backward(); opt2.step()
print("\nKhoi tao khac -> nghiem khac:")
print("A=%.4f B=%.4f C=%.4f D=%.4f | sqerr=%.8f" % (A2.item(), B2.item(), C2.item(), D2.item(), e.item()))
```

**Yêu cầu 1: Tensor X 24 phần tử, reshape 4×6, 2×12, 2×3×4**

```python
import torch

X = torch.arange(24)
print("X:", X)
print("shape:", X.shape, "| numel:", X.numel())

print("\nReshape 4 x 6:")
print(X.reshape(4, 6))
print("\nReshape 2 x 12:")
print(X.reshape(2, 12))
print("\nReshape 2 x 3 x 4:")
print(X.reshape(2, 3, 4))

print("\nKiem tra so phan tu sau moi lan reshape:",
      X.reshape(4, 6).numel(), X.reshape(2, 12).numel(), X.reshape(2, 3, 4).numel())
```

**Nhận xét:** Mọi cách reshape đều giữ nguyên 24 phần tử vì 4×6 = 2×12 = 2×3×4 = 24.
Reshape chỉ thay đổi cách "nhìn" dữ liệu chứ không sao chép hay thay đổi giá trị.

**Yêu cầu 2: Hai Tensor A, B kích thước 3 × 4 — tính A+B, A\*B, A/B, trung bình và độ lệch chuẩn theo cột**

```python
import torch

A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("A:"); print(A)
print("\nB:"); print(B)
print("\nA + B:"); print(A + B)
print("\nA * B (element-wise):"); print(A * B)
print("\nA / B:"); print(A / B)
print("\nTrung binh theo cot  - mean(dim=0):"); print(torch.mean(A, dim=0))
print("\nDo lech chuan theo cot - std(dim=0):"); print(torch.std(A, dim=0))
```

**Nhận xét:** `A+B`, `A*B`, `A/B` đều là phép toán theo từng phần tử nên kết quả giữ shape `[3, 4]`.
`mean(dim=0)` và `std(dim=0)` thu gọn chiều hàng nên trả về vector 4 phần tử — mỗi phần tử ứng với một cột.

**Yêu cầu 3: Cho y = 2x² + 5x + 3 — dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng tính thủ công**

```python
import torch

x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

manual = 4 * 2 + 5        # dy/dx = 4x + 5
print("y(2)                       =", y.item())
print("Autograd  dy/dx tai x = 2  =", x.grad.item())
print("Thu cong  dy/dx = 4x + 5   =", manual)
print("Bang nhau?", abs(manual - x.grad.item()) < 1e-6)
```

**Nhận xét:** y = 2x² + 5x + 3 ⟹ dy/dx = 4x + 5. Tại x = 2: dy/dx = 4(2) + 5 = **13**,
đúng bằng giá trị `x.grad` do autograd tính ra.

**Yêu cầu 4: Thay đa thức trong Bài 9 bằng y = 2x² − 3x + 5 — huấn luyện và cho biết các hệ số tìm được**

```python
import numpy as np
import torch

polynomial = np.poly1d([2, -3, 5])        # y = 2x^2 - 3x + 5
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y,  dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:", w.detach().numpy().ravel())

for epoch in range(5000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    if (epoch + 1) % 1000 == 0:
        print("Epoch %5d | MSE = %.8f" % (epoch + 1, mse.item()))

learned = w.detach().numpy().ravel()
true    = np.array([2.0, -3.0, 5.0])
print("\n%-16s %12s %12s %12s" % ("He so", "That", "Hoc duoc", "Sai so"))
for n, t, l in zip(["a (x^2)", "b (x)", "c (hang so)"], true, learned):
    print("%-16s %12.4f %12.4f %12.6f" % (n, t, l, abs(t - l)))
```

**Nhận xét:** Sau 5000 epoch, ba hệ số học được hội tụ rất gần [2, −3, 5] và MSE giảm về xấp xỉ 0.
So với Bài 9, ở đây cần nhiều epoch hơn vì hệ số cần học có biên độ lớn hơn và xuất phát điểm ngẫu nhiên
có thể xa nghiệm hơn. Điều này minh hoạ: cùng một mô hình tuyến tính theo tham số, Gradient Descent + autograd
có thể khôi phục lại chính xác các hệ số của đa thức sinh dữ liệu.
